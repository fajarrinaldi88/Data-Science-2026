```python
#import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

#load dataset iris
df = sns.load_dataset('iris')

#menampilkan jumlah baris dan kolom
print('Shape:', df.shape)
#menampilkan tipe data tiap kolom
print(df.dtypes)
#menampilkan 5 data pertama
print(df.head())
#menampilkan ringkasan statistik
print(df.describe().round(3))
```

```python
for col_name in df.select_dtypes(include='number').columns:
    col = df[col_name]
    print(f'\n=== {col_name} ===')
    print(f'Mean      : {col.mean():.3f}')
    print(f'Median    : {col.median():.3f}')
    print(f'Std Dev   : {col.std():.3f}')
    print(f'Varians   : {col.var():.3f}')
    print(f'Skewness  : {col.skew():.3f}')
    print(f'Kurtosis  : {col.kurt():.3f}')
```

```python
fig, ax = plt.subplots(figsize=(8,4))

sns.histplot(
    df['sepal_length'],
    kde=True,
    color='#028090',
    bins=20,
    ax=ax
)

ax.axvline(
    df['sepal_length'].mean(),
    color='red',
    linestyle='--',
    label=f"Mean = {df['sepal_length'].mean():.2f}"
)

ax.axvline(
    df['sepal_length'].median(),
    color='orange',
    linestyle='--',
    label=f"Median = {df['sepal_length'].median():.2f}"
)

ax.set_title('Distribusi Sepal Length')
ax.legend()

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

sns.boxplot(
    data=df,
    x='species',
    y='petal_length',
    palette='Set2',
    ax=axes[0]
)

axes[0].set_title('Boxplot Petal Length')

sns.violinplot(
    data=df,
    x='species',
    y='petal_length',
    palette='Set2',
    inner='box',
    ax=axes[1]
)

axes[1].set_title('Violin Plot Petal Length')

plt.tight_layout()
plt.show()
```

```python
# Menghitung korelasi Pearson
corr = df.drop('species', axis=1).corr(method='pearson')
print(corr.round(3))

# Mencari pasangan korelasi tertinggi
mask = np.triu(np.ones(corr.shape, dtype=bool))
corr_masked = corr.where(~mask)
max_pair = corr_masked.stack().idxmax()
print(f'\nKorelasi tertinggi: {max_pair} = {corr.loc[max_pair]:.3f}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14,5))

sns.scatterplot(
    data=df,
    x='petal_length',
    y='petal_width',
    hue='species',
    palette='Set2',
    ax=axes[0]
)

sns.regplot(
    data=df,
    x='petal_length',
    y='petal_width',
    scatter=False,
    color='gray',
    ax=axes[0]
)

axes[0].set_title('Scatter Plot')

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='RdBu_r',
    vmin=-1,
    vmax=1,
    square=True,
    ax=axes[1]
)

axes[1].set_title('Heatmap Korelasi')

plt.tight_layout()
plt.show()
```

## Kesimpulan
Dari analisis eksplorasi yang telah dilakukan terhadap dataset Iris, beberapa temuan utama dapat disimpulkan:

*   **Distribusi Fitur:** Fitur `sepal_length` menunjukkan distribusi yang mendekati normal, dengan nilai rata-rata dan median yang berdekatan. Sementara itu, fitur `petal_length` dan `petal_width` memiliki distribusi yang cenderung terbagi, dengan tiga kelompok yang jelas sesuai dengan tiga spesies Iris.

*   **Perbedaan Antar Spesies:** Boxplot dan Violin Plot `petal_length` dengan jelas menunjukkan perbedaan signifikan antara spesies 'setosa', 'versicolor', dan 'virginica'. Spesies 'setosa' memiliki `petal_length` terpendek dan paling konsisten, sedangkan 'virginica' memiliki yang terpanjang.

*   **Korelasi Antar Fitur:** Terdapat korelasi positif yang sangat kuat antara `petal_length` dan `petal_width` (0.963), menunjukkan bahwa kedua fitur ini sangat terkait. Korelasi positif yang kuat juga terlihat antara `sepal_length` dengan `petal_length` (0.872) dan `petal_width` (0.818).
