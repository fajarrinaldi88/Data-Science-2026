```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a style for the plots for better aesthetics
sns.set_style("whitegrid")
```

### Visualisasi Data dari Dataset Tips

Dataset 'tips' berisi informasi transaksi di sebuah restoran, termasuk total tagihan, jumlah tip, jenis kelamin pelanggan, apakah mereka perokok atau tidak, hari dalam seminggu, waktu makan (makan siang atau malam), dan jumlah orang dalam rombongan.

### 1. Bar Chart: Jumlah Perokok vs. Bukan Perokok per Hari

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=tips, x='day', hue='smoker', palette='viridis', order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Jumlah Pelanggan Perokok dan Bukan Perokok per Hari', fontsize=16)
plt.xlabel('Hari', fontsize=12)
plt.ylabel('Jumlah Pelanggan', fontsize=12)
plt.legend(title='Perokok', loc='upper right')
plt.tight_layout()
plt.show()
```

**What?** Grafik batang ini menunjukkan jumlah pelanggan perokok dan bukan perokok yang berkunjung ke restoran pada setiap hari dalam seminggu. Terlihat bahwa pada hari Sabtu dan Minggu, jumlah pelanggan, baik perokok maupun bukan perokok, cenderung lebih tinggi dibandingkan hari Kamis dan Jumat. Jumlah bukan perokok selalu lebih banyak daripada perokok di semua hari, dengan puncaknya pada hari Minggu (57 bukan perokok).

**So what?** Informasi ini penting karena menunjukkan pola kunjungan pelanggan berdasarkan hari dan kebiasaan merokok mereka. Restoran dapat menggunakan data ini untuk mengatur staf, persediaan, atau merancang promosi khusus pada hari-hari tertentu. Misalnya, pada akhir pekan, restoran mungkin membutuhkan lebih banyak staf.

**Now what?** Kita bisa melanjutkan dengan menganalisis rata-rata pengeluaran atau tip yang diberikan oleh kelompok perokok dan bukan perokok pada hari-hari tertentu untuk memahami lebih lanjut nilai ekonomi dari masing-masing segmen pelanggan.

### 2. Histogram: Distribusi Total Tagihan

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=tips, x='total_bill', bins=15, kde=True, color='skyblue')
plt.title('Distribusi Total Tagihan', fontsize=16)
plt.xlabel('Total Tagihan ($)', fontsize=12)
plt.ylabel('Frekuensi', fontsize=12)
plt.tight_layout()
plt.show()
```

**What?** Histogram ini menampilkan distribusi total tagihan di restoran. Mayoritas transaksi memiliki total tagihan antara $10 hingga $25, dengan puncaknya sekitar $15-$20. Distribusi ini terlihat miring ke kanan (positively skewed), menunjukkan bahwa ada beberapa transaksi dengan total tagihan yang sangat tinggi, tetapi jumlahnya tidak sebanyak transaksi dengan total tagihan yang lebih rendah.

**So what?** Memahami distribusi total tagihan membantu restoran dalam penetapan harga, strategi promosi, dan perkiraan pendapatan. Misalnya, karena sebagian besar tagihan di bawah $25, restoran dapat membuat menu khusus atau paket makan siang yang menarik untuk segmen ini.

**Now what?** Akan menarik untuk melihat apakah ada perbedaan distribusi total tagihan berdasarkan waktu makan (makan siang/malam) atau ukuran rombongan pelanggan. Ini bisa memberikan wawasan lebih lanjut tentang kebiasaan pengeluaran pelanggan.

### 3. Boxplot: Total Tagihan berdasarkan Hari

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=tips, x='day', y='total_bill', palette='pastel', order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Total Tagihan per Hari', fontsize=16)
plt.xlabel('Hari', fontsize=12)
plt.ylabel('Total Tagihan ($)', fontsize=12)
plt.tight_layout()
plt.show()
```

**What?** Boxplot ini menunjukkan sebaran total tagihan untuk setiap hari dalam seminggu. Terlihat bahwa total tagihan rata-rata (garis tengah dalam kotak) dan rentang interkuartil (IQR) pada hari Sabtu dan Minggu lebih tinggi dibandingkan hari Kamis dan Jumat. Hari Sabtu memiliki median total tagihan sekitar $20, dan hari Minggu sekitar $21. Terdapat juga beberapa outlier, terutama pada hari Minggu, dengan total tagihan yang jauh lebih tinggi dari sebagian besar transaksi ($50+).

**So what?** Boxplot ini mengkonfirmasi bahwa akhir pekan (Sabtu dan Minggu) adalah waktu di mana pelanggan cenderung memiliki total tagihan yang lebih tinggi. Ini penting untuk perencanaan keuangan dan operasional, seperti memprediksi pendapatan harian dan mengoptimalkan persediaan untuk menu-menu yang lebih mahal pada akhir pekan.

**Now what?** Kita bisa menyelidiki lebih lanjut outlier tersebut. Siapa pelanggan yang membuat tagihan tinggi ini? Apakah mereka rombongan besar, atau individu yang memesan hidangan mahal? Informasi ini dapat membantu restoran menargetkan segmen pelanggan bernilai tinggi.

### 4. Scatter Plot: Hubungan antara Total Tagihan dan Tip

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='smoker', size='size', sizes=(20, 400), alpha=0.7)
plt.title('Hubungan antara Total Tagihan dan Tip', fontsize=16)
plt.xlabel('Total Tagihan ($)', fontsize=12)
plt.ylabel('Tip ($)', fontsize=12)
plt.legend(title='Perokok', loc='upper left')
plt.tight_layout()
plt.show()
```

**What?** Scatter plot ini menggambarkan hubungan antara total tagihan dan jumlah tip yang diberikan. Secara umum, terlihat ada korelasi positif: semakin tinggi total tagihan, semakin besar juga tip yang diberikan. Titik-titik diplot berdasarkan status perokok (warna) dan ukuran rombongan (ukuran titik). Tampak bahwa pelanggan bukan perokok (warna hijau) cenderung memiliki total tagihan dan tip yang lebih tersebar, sedangkan perokok (warna kuning) juga menunjukkan pola serupa namun dengan lebih sedikit data.

**So what?** Korelasi positif antara total tagihan dan tip sangat intuitif dan penting bagi karyawan yang mengandalkan tip sebagai bagian dari pendapatan mereka. Restoran dapat mendorong peningkatan ukuran tagihan (misalnya, melalui penawaran hidangan penutup) untuk secara tidak langsung meningkatkan potensi tip bagi staf. Informasi mengenai ukuran rombongan juga membantu memahami dinamika tip.

**Now what?** Analisis lebih lanjut bisa mencakup perbandingan persentase tip dari total tagihan berdasarkan berbagai kategori (hari, waktu, jenis kelamin, perokok/bukan perokok) untuk mengidentifikasi faktor-faktor yang paling mempengaruhi kemurahan hati pelanggan dalam memberikan tip.
